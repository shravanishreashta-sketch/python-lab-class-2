```python
#q6_simple calculator function
 # Define calculator operations
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error! Division by zero."
    return a / b

# User input
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Choose operation (+, -, *, /): ")

# Execute choice
if operation == '+':
    print("Result:", add(num1, num2))
elif operation == '-':
    print("Result:", subtract(num1, num2))
elif operation == '*':
    print("Result:", multiply(num1, num2))
elif operation == '/':
    print("Result:", divide(num1, num2))
else:
    print("Invalid operation chosen.")
```

```python
#Q8-largest among 3 nos
def find_largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

# Example usage:
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

largest = find_largest(num1, num2, num3)
print(f"The largest number is {largest}")
```

```python
#Q9-factorial using function
def calculate_factorial(n):
    if n < 0:
        return "Factorial does not exist for negative numbers"
    elif n == 0 or n == 1:
        return 1
    else:
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result

# Example usage:
num = int(input("Enter a positive integer: "))
fact = calculate_factorial(num)
print(f"The factorial of {num} is {fact}")
```

```python
#Q10-prime no checker
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage:
num = int(input("Enter a number to check: "))
if is_prime(num):
    print(f"{num} is a prime number (True).")
else:
    print(f"{num} is not a prime number (False).")
```

```python
#q11-attendance system
# Initialize an empty list to store student names
students_list = []

def add_student():
    name = input("Enter student name to add: ").strip()
    if name:
        students_list.append(name)
        print(f"Success: '{name}' added.")
    else:
        print("Error: Name cannot be empty.")

def remove_student():
    name = input("Enter student name to remove: ").strip()
    if name in students_list:
        students_list.remove(name)
        print(f"Success: '{name}' removed.")
    else:
        print(f"Error: '{name}' not found in the list.")

def display_all_students():
    if not students_list:
        print("\nThe attendance list is currently empty.")
    else:
        print("\n--- Current Attendance List ---")
        for index, student in enumerate(students_list, start=1):
            print(f"{index}. {student}")
        print("--------------------------------")

# Interactive Menu Loop
while True:
    print("\n=== Attendance System Menu ===")
    print("1. Add Student")
    print("2. Remove Student")
    print("3. Display All Students")
    print("4. Exit")

    choice = input("Select an option (1-4): ").strip()

    if choice == '1':
        add_student()
    elif choice == '2':
        remove_student()
    elif choice == '3':
        display_all_students()
    elif choice == '4':
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid choice! Please select a valid number between 1 and 4.")
```

```python
#q12-employee salary dictionary
# Initialize an empty dictionary to store employee details
employee_db = {}

def add_employee():
    name = input("Enter employee name to add: ").strip()
    if not name:
        print("Error: Name cannot be empty.")
        return
    try:
        salary = float(input(f"Enter salary for {name}: "))
        employee_db[name] = salary
        print(f"Success: Added {name} with a salary of {salary}.")
    except ValueError:
        print("Error: Invalid salary amount. Please enter a number.")

def update_salary():
    name = input("Enter employee name to update salary: ").strip()
    if name in employee_db:
        try:
            new_salary = float(input(f"Enter new salary for {name}: "))
            employee_db[name] = new_salary
            print(f"Success: Updated salary for {name} to {new_salary}.")
        except ValueError:
            print("Error: Invalid salary amount. Please enter a number.")
    else:
        print(f"Error: Employee '{name}' does not exist.")

def search_employee():
    name = input("Enter employee name to search: ").strip()
    if name in employee_db:
        print(f"\n--- Employee Details ---")
        print(f"Name: {name}")
        print(f"Salary: ${employee_db[name]:,.2f}")
        print("------------------------")
    else:
        print(f"Error: Employee '{name}' not found.")

def display_all_employees():
    if not employee_db:
        print("\nThe employee database is currently empty.")
    else:
        print("\n--- Current Employee Directory ---")
        for name, salary in employee_db.items():
            print(f"Name: {name:<15} | Salary: ${salary:,.2f}")
        print("----------------------------------")

# Interactive Menu Loop
while True:
    print("\n=== Employee Salary System Menu ===")
    print("1. Add Employee")
    print("2. Update Salary")
    print("3. Search Employee")
    print("4. Display All Employees")
    print("5. Exit")

    choice = input("Select an option (1-5): ").strip()

    if choice == '1':
        add_employee()
    elif choice == '2':
        update_salary()
    elif choice == '3':
        search_employee()
    elif choice == '4':
        display_all_employees()
    elif choice == '5':
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid choice! Please select a valid number between 1 and 5.")
```

```python
#q14-voting eligibility checker
def check_voting_eligibility(ages_tuple):
    print("\n--- Voting Eligibility Results ---")
    for index, age in enumerate(ages_tuple, start=1):
        status = "Eligible" if age >= 18 else "Not Eligible (Under 18)"
        print(f"Person {index}: Age {age} -> {status}")
    print("---------------------------------")

# Main Program Execution
try:
    num_people = int(input("How many ages do you want to check? "))
    ages_list = []

    for i in range(num_people):
        age = int(input(f"Enter age for person {i+1}: "))
        ages_list.append(age)

    # Convert to a tuple as required by the question
    final_ages = tuple(ages_list)

    # Check eligibility
    check_voting_eligibility(final_ages)

except ValueError:
    print("Error: Please enter valid numbers for age.")
```

```python
#q15-word frequency counter
# Take input string from user
sentence = input("Enter a sentence: ")

# Convert to lowercase and split into a list of words
words = sentence.lower().split()

# Initialize empty dictionary to track counts
frequency_dict = {}

# Count the frequency of each word
for word in words:
    # Clean basic punctuation marks if present
    word = word.strip(".,!?\"'")
    if word:
        frequency_dict[word] = frequency_dict.get(word, 0) + 1

# Display final frequency dictionary
print("\nOutput Word Frequency Dictionary:")
print(frequency_dict)
```

```python
#q16-library book system
# Function to add a book to the library
def add_book(library):
    book = input("Enter the name of the book to add: ").strip()
    if book:
        library.append(book)
        print(f"'{book}' has been successfully added.")
    else:
        print("Book name cannot be empty.")

# Function to remove a book from the library
def remove_book(library):
    book = input("Enter the name of the book to remove: ").strip()
    if book in library:
        library.remove(book)
        print(f"'{book}' has been successfully removed.")
    else:
        print(f"'{book}' not found in the library.")

# Function to search for a book in the library
def search_book(library):
    book = input("Enter the name of the book to search: ").strip()
    if book in library:
        print(f"Yes, '{book}' is available in the library.")
    else:
        print(f"No, '{book}' is not available.")

# Function to display all books in the library
def display_books(library):
    if not library:
        print("The library is currently empty.")
    else:
        print("\n--- Books in Library ---")
        for index, book in enumerate(library, 1):
            print(f"{index}. {book}")

# Main function to handle the menu-driven loop
def main():
    library_books = []  # List data structure to store books

    while True:
        print("\n=== Library Book System ===")
        print("1. Add Book")
        print("2. Remove Book")
        print("3. Search Book")
        print("4. Display Books")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()

        if choice == '1':
            add_book(library_books)
        elif choice == '2':
            remove_book(library_books)
        elif choice == '3':
            search_book(library_books)
        elif choice == '4':
            display_books(library_books)
        elif choice == '5':
            print("Exiting the program. Goodbye!")
            break
        else:
            print("Invalid choice! Please select an option between 1 and 5.")

# Run the program
if __name__ == "__main__":
    main()
```

```python
#q17-student marks analyser
# Function to calculate and display the results
def analyze_results(marks_list):
    # 1. Find Average
    total_marks = sum(marks_list)
    average = total_marks / len(marks_list)

    # 2. Find Highest and Lowest
    highest = max(marks_list)
    lowest = min(marks_list)

    # 3. Find Passed students (>40)
    # Using a list comprehension to filter marks strictly greater than 40
    passed_students = [marks for marks in marks_list if marks > 40]
    passed_count = len(passed_students)

    # Display the analysis
    print("\n--- Result Analysis ---")
    print(f"Average Marks: {average:.2f}")
    print(f"Highest Marks: {highest}")
    print(f"Lowest Marks: {lowest}")
    print(f"Number of Passed Students (>40): {passed_count} (Marks: {passed_students})")

# Main function to get user input
def main():
    print("=== Student Result Analyzer ===")
    marks_list = []

    # Input marks for 5 students
    for i in range(1, 6):
        while True:
            try:
                marks = float(input(f"Enter marks for student {i}: "))
                if 0 <= marks <= 100:  # Validating realistic marks range
                    marks_list.append(marks)
                    break
                else:
                    print("Please enter marks between 0 and 100.")
            except ValueError:
                print("Invalid input! Please enter a numerical value.")

    # Call the analyzer function
    analyze_results(marks_list)

# Run the program
if __name__ == "__main__":
    main()
```

```python
#q18-password strength checker
# Function to check the strength of a password
def check_password_strength(password):
    # Initialize flag variables as False
    has_digit = False
    has_upper = False
    has_lower = False

    # Check each character in the password
    for char in password:
        if char.isdigit():
            has_digit = True
        elif char.isupper():
            has_upper = True
        elif char.islower():
            has_lower = True

    # Check minimum length criteria
    has_min_length = len(password) >= 8

    # Display evaluation results
    print("\n--- Password Criteria Check ---")
    print(f"Contains digit: {'✔️ Yes' if has_digit else '❌ No'}")
    print(f"Contains uppercase letter: {'✔️ Yes' if has_upper else '❌ No'}")
    print(f"Contains lowercase letter: {'✔️ Yes' if has_lower else '❌ No'}")
    print(f"Minimum length of 8 characters: {'✔️ Yes' if has_min_length else '❌ No'}")

    # Final verdict
    if has_digit and has_upper and has_lower and has_min_length:
        print("\nVerdict: Strong Password! 💪")
    else:
        print("\nVerdict: Weak Password! ⚠️ Please meet all criteria.")

# Main function to run the application
def main():
    print("=== Password Strength Checker ===")
    user_password = input("Enter a password to test: ")
    check_password_strength(user_password)

if __name__ == "__main__":
    main()
```

```python
#q19-contack book
# Function to add a contact
def add_contact(contact_book):
    name = input("Enter contact name: ").strip()
    if not name:
        print("Name cannot be empty.")
        return

    if name in contact_book:
        print(f"'{name}' already exists with number: {contact_book[name]}")
        overwrite = input("Do you want to overwrite it? (yes/no): ").lower()
        if overwrite != 'yes':
            return

    phone = input("Enter phone number: ").strip()
    contact_book[name] = phone
    print(f"Contact '{name}' added successfully.")

# Function to search for a contact
def search_contact(contact_book):
    name = input("Enter the name to search: ").strip()
    if name in contact_book:
        print(f"Found: {name} -> {contact_book[name]}")
    else:
        print(f"Contact '{name}' not found.")

# Function to delete a contact
def delete_contact(contact_book):
    name = input("Enter the name to delete: ").strip()
    if name in contact_book:
        del contact_book[name]
        print(f"Contact '{name}' deleted successfully.")
    else:
        print(f"Contact '{name}' not found.")

# Function to display all contacts
def display_contacts(contact_book):
    if not contact_book:
        print("Your contact book is empty.")
    else:
        print("\n--- Contact List ---")
        for name, phone in contact_book.items():
            print(f"{name}: {phone}")

# Main function for the menu loop
def main():
    contact_book = {}  # Dictionary to store contacts

    while True:
        print("\n=== Contact Book Application ===")
        print("1. Add Contact")
        print("2. Search Contact")
        print("3. Delete Contact")
        print("4. Display Contacts")
        print("5. Exit")

        choice = input("Enter choice (1-5): ").strip()

        if choice == '1':
            add_contact(contact_book)
        elif choice == '2':
            search_contact(contact_book)
        elif choice == '3':
            delete_contact(contact_book)
        elif choice == '4':
            display_contacts(contact_book)
        elif choice == '5':
            print("Exiting application. Goodbye!")
            break
        else:
            print("Invalid choice! Please enter a number between 1 and 5.")

if __name__ == "__main__":
    main()
```

```python
#q20-hacker rank style problem
# Given list
numbers = [10, 20, 30, 40, 50]

# 1. Find Maximum without built-in functions
max_val = numbers[0]
for num in numbers:
    if num > max_val:
        max_val = num

# 2. Find Minimum without built-in functions
min_val = numbers[0]
for num in numbers:
    if num < min_val:
        min_val = num

# 3. Reverse List without built-in functions
reversed_list = []
# Loop backwards through the list indices
for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

# Display Results
print("Original List:", numbers)
print("Maximum Value:", max_val)
print("Minimum Value:", min_val)
print("Reversed List:", reversed_list)
```

```python
#q21 Initialize global balance variable
balance = 1000.0

def deposit(amount):
    global balance
    if amount > 0:
        balance += amount
        print(f"Successfully deposited ${amount:.2f}")
    else:
        print("Deposit amount must be positive.")

def withdraw(amount):
    global balance
    if amount <= 0:
        print("Withdrawal amount must be positive.")
    elif amount > balance:
        print("Insufficient balance!")
    else:
        balance -= amount
        print(f"Successfully withdrew ${amount:.2f}")

def check_balance():
    global balance
    print(f"Your current balance is: ${balance:.2f}")

# Example Usage:
check_balance()
deposit(500)
check_balance()
withdraw(300)
check_balance()
withdraw(2000) # Testing insufficient funds
```
