```python
#list with different data types
my_list=[10,3.5,"hello",True]
print("list:",my_list)
print("type of list:",type(my_list))
for i in my_list:
    print(i,">",type(i))
```

```python
#list operations
a=[1,2,3]
b=[4,5,6]
#concatenation
print("conactenation:",a+b)
print("repetition:",a*3)
print("Is 2 in list a?",2 in a)
print("length of list a:",len(a))
```

```python
nums=[10,5,20,8,15]
print("max:",max(nums))
print("min:",min(nums))
print("sum:",sum(nums))
nums.sort()
print("ascending order:",nums)
nums.sort(reverse=True)
print("descending order:",nums)
print("index of 8:",nums.index(8))
```

```python
#list modification
lst=[10,20,30]
lst.append(40)
print("after append:",lst)
lst.insert(1,15)
print("after insert:",lst)
lst.remove(20)
print("after remove:",lst)
lst.pop()
print("after pop:",lst)
lst.reverse()
print("after reverse:",lst)
```
