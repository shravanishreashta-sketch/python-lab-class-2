```python
#type of triangle

a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))

# First check if triangle is valid
if (a + b > c) and (a + c > b) and (b + c > a):

    if a == b == c:
        print("Equilateral")
    elif a == b or b == c or a == c:
        print("Isosceles")
    else:
        print("Scalene")
else:
    print("Invalid triangle")
```

```python
#electricity bill

units = float(input("Enter units consumed: "))

bill = 0

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)
else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print("Total bill: ₹", bill)
```

```python
# Salary bonus calculator

salary = float(input("Enter your salary: "))
years = int(input("Enter years of service: "))

if years >= 5:
    bonus = salary * 0.10
else:
    bonus = salary * 0.05

final_salary = salary + bonus

print("Bonus: ₹", bonus)
print("Final Salary: ₹", final_salary)
```

```python
# ATM Withdrawal Program

balance = float(input("Enter account balance: "))
withdraw = float(input("Enter withdrawal amount: "))

if withdraw % 100 != 0:
    print("Transaction Failed")
elif withdraw > balance:
    print("Transaction Failed")
elif (balance - withdraw) < 500:
    print("Transaction Failed")
else:
    print("Transaction Successful")
```

```python
# quadratic equation roots nature

a = float(input("Enter value of a: "))
b = float(input("Enter value of b: "))
c = float(input("Enter value of c: "))

D = (b ** 2) - (4 * a * c)

print("Discriminant (D) =", D)

if D > 0:
    print("Real and Distinct")
elif D == 0:
    print("Real and Equal")
else:
    print("Imaginary")
```

```python
# Discount Calculator

amount = float(input("Enter purchase amount: "))

if amount >= 5000:
    discount = amount * 0.20
elif amount >= 2000:
    discount = amount * 0.10
else:
    discount = 0

final_amount = amount - discount

print("Discount: ₹", discount)
print("Final Amount: ₹", final_amount)
```

```python
# Menu Driven Program

print("1.check even/odd")
print("2.check positive/negative")
print("3.exit")
choice = int(input("Enter your choice: "))

if choice == 1:
        num = int(input("Enter a number: "))
        if num % 2 == 0:
            print("Even number")
        else:
            print("Odd number")

elif choice == 2:
        num = float(input("Enter a number: "))
        if num > 0:
            print("Positive number")
        elif num < 0:
            print("Negative number")
        else:
            print("Zero")

elif choice == 3:
        print("Exiting program...")


else:
        print("Invalid choice. Please try again.")
```
